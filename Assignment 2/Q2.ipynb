```python
import numpy as np
with open("sensor_data.csv", "r") as file:
    lines = file.readlines()  

data_list = [line.strip().split(",") for line in lines]
data = np.array(data_list)

```

```python
data = np.array(data_list)
data[data == "-999"] = np.nan

data = data.astype(float)

data[(data < 0) | (data > 100)] = np.nan
```

```python
mean_per_sensor = np.nanmean(data, axis=0)
print("Mean per sensor (first 5):", mean_per_sensor)

median_per_hour = np.nanmedian(data, axis=1)
print("Median per hour (first 5):", median_per_hour[:5])

nan_count_per_sensor = np.isnan(data).sum(axis=0)
worst_sensor = np.argmax(nan_count_per_sensor)

print("Sensor with most invalid readings:", worst_sensor)
print("Invalid count:", nan_count_per_sensor[worst_sensor])
```

```python
Xmin = np.nanmin(data)
Xmax = np.nanmax(data)
Xnorm = (data - Xmin) / (Xmax - Xmin)
print(Xnorm[:5])

np.savetxt("sensor_data_normalized_csv", Xnorm)
```
