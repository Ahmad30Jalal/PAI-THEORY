```python
import pandas as pd
df1 = pd.read_csv("titanic_cleaned.csv")
df2 = pd.read_csv("ticket_fares.csv")
```

```python
df3 = pd.merge(df1, df2, on='Ticket')
df3
```

```python
age_labels = ['Child', 'Teen', 'Adult', 'Senior']
age_bins = [0, 16, 25, 50, 100]
df3['AgeGroup'] = pd.cut(df3['Age'], bins = age_bins, labels = age_labels, right = True)
survival_rate = df3.groupby(['Sex', 'AgeGroup'], observed= True)['Survived'].mean()
df3
print(survival_rate)
```

```python
with open('report.txt', 'w') as f:
    f.write("The anaylsis shows that women and children had higher survival rate than men." \
    "The survival rate of women ranges from 47.8% for child to 93.3% for seniors. On the other" \
    "hand the survival rate for men ranges from 10% for seniors to 25.3% for children. However " \
    "in females the children had the least survival rate compared to men where children had the " \
    "higest survival rate. The data supports the Women first hypothesis but it not supports children" \
    "first for female category")
```

```python
class_survival = df3.groupby("Pclass")['Survived'].mean()
print(class_survival)
fare_labels = ['Low', 'Medium', 'High', 'Very High']
df3['FareBin'] = pd.qcut(df3['Fare_x'], 4, labels = fare_labels)
fare_survival = df3.groupby('FareBin', observed=True)['Survived'].mean()
print(fare_survival)
df3
```
